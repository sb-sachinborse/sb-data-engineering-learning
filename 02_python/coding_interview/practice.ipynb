```python
import time

class ETL:

    #constructor
    def __init__(self):
        print("ETL Pipeline Initialized")
        time.sleep(2)

    #methods
    def extraction(self):
        print("extraction starting")
        time.sleep(5)

    def transformation(self):
        print("transformation starting.....")
        time.sleep(5)

    def load(self):
        print("loading data....")
        time.sleep(5)
        print('DONE')

    #abstraction
    def run(self):
        self.extraction()
        self.transformation()
        self.load()

def main():
    etl = ETL() 
    etl.run()


if __name__ == "__main__":
    main()
```
