## 🐍 Python OOP for Data Engineers 

#### 1️⃣ Why OOP for Data Engineers?
Without OOP (Bad)
- Hard-coded logic
- Repeated code
-Difficult to maintain pipelines

With OOP (Good)
- Reusable ETL component
- Clear separation of concerns
- Easy scaling & testing

#### 2️⃣ Class & Object (Foundation)

- Class → Blueprint
- Object → Instance of class


```python
class DataEngineer:
    def work(self):
        print("Building data pipelines")

de = DataEngineer()
de.work()
```

#### 3️⃣ Constructor (__init__)
🔹 Used to initialize object data

```python
class Employee:
    def __init__(self, name, role):
        self.name = name
        self.role = role

    def show(self):
        print(self.name, self.role)

e1 = Employee("Krishna", "Data Engineer")
e1.show()
```

#### 4️⃣ Instance Variables vs Methods

```python
class Job:
    def __init__(self, company):
        self.company = company   # instance variable

    def display(self):           # instance method
        print(self.company)
```

#### 5️⃣ Encapsulation (Data Hiding)
Protect internal data using access control

✔ Prevents accidental modification

```python
class DatabaseConfig:
    def __init__(self):
        self.__password = "secret123"

    def get_password(self):
        return self.__password
```

#### 6️⃣ Inheritance (Code Reuse)
Child class inherits parent functionality.
- 🧠 Used for multiple data sources

```python
class DataSource:
    def connect(self):
        print("Connecting to source")

class MySQLSource(DataSource):
    def read(self):
        print("Reading MySQL data")

db = MySQLSource()
db.connect()
db.read()
```

#### 7️⃣ Method Overriding (Polymorphism)
✔ Same method, different behavior

```python
class DataSource:
    def read(self):
        print("Reading generic data")

class S3Source(DataSource):
    def read(self):
        print("Reading data from S3")
```

#### 8️⃣ Polymorphism (Real ETL Pattern)
🧠 Spark, Airflow operators use this heavily

```python
def load_data(source):
    source.read()

load_data(MySQLSource())
load_data(S3Source())
```

#### 9️⃣ Abstraction (Interface-Like Design)
🔹 Using abc module

```python
from abc import ABC, abstractmethod

class DataSource(ABC):
    @abstractmethod
    def read(self):
        pass
```

```python
class APISource(DataSource):
    def read(self):
        print("Reading API data")
```

#### 🔟 Composition (Better than Inheritance)
🧠 Preferred in production pipelines

```python
class Logger:
    def log(self, msg):
        print(msg)

class ETLJob:
    def __init__(self):
        self.logger = Logger()

    def run(self):
        self.logger.log("ETL started")
```

#### 1️⃣1️⃣ Class Variables & Static Methods

```python
class Pipeline:
    version = "1.0"

    @staticmethod
    def info():
        print("Data Pipeline")
```

#### 1️⃣2️⃣ Property Decorators (Professional)

```python
class Employee:
    def __init__(self, salary):
        self._salary = salary

    @property
    def salary(self):
        return self._salary
```

#### 1️⃣3️⃣ Magic Methods (__str__, __repr__)

```python
class Job:
    def __str__(self):
        return "Data Engineer Job"

print(Job())
```

### 1️⃣4️⃣ Real-World Mini ETL (OOP Style)

```python
class Extract:
    def get_data(self):
        return ["10", "20", None, "30"]

class Transform:
    def clean(self, data):
        return [int(x) for x in data if x]

class Load:
    def save(self, data):
        print("Saving:", data)

class ETL:
    def __init__(self):
        self.e = Extract()
        self.t = Transform()
        self.l = Load()

    def run(self):
        data = self.e.get_data()
        data = self.t.clean(data)
        self.l.save(data)

ETL().run()
```
