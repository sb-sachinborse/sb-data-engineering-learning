```python
#👉 Find the duplicate elements from list

List = [1,2,2,-1,5,6,5]

#best approch 
# ⏱️ Time: O(n)
# 💾 Space: O(n)
def find_duplicate(list):

    seen = set()
    duplicate = []

    for element in list:
        if element in seen and element not in duplicate:
            duplicate.append(element)
        else:
            seen.add(element)
    return duplicate

find_duplicate(List)


#second approch
# def duplicates_value(list):
#     unique = []

#     for element in list:
#         if list.count(element) > 1:
#             if element not in unique:
#                 unique.append(element)
#     return unique

# duplicates_value(List)


#third approch
# from collections import Counter

# def find_duplicates(list):
#     freq = Counter(list)
#     return [list for list,count in freq.items() if count > 1]

# find_duplicates(List)

# 🎯 When to Use Counter (Interview Tip)

# ✔ Data fits in memory
# ✔ Need frequency info
# ✔ Clean & readable code

# ❌ Avoid if:
# Streaming data
# Very memory-constrained environments

```

```python
# 👉 find the duplicate_transaction
transaction = [101,303,405,101,904,402,784,904,103,433]

seen = set()
duplicates = set()

for t in transaction:
    if t in seen:
        duplicates.add(t)
    else:
        seen.add(t)

print(list(duplicates))
```

```python
# 👉 Find duplicate values AND their original indices

transaction = [101,303,405,101,904,402,784,904,103,433]

# output:
# 101 -> [0, 3]
# 904 -> [4, 7]

def find_duplicate_indices(transaction: list) -> dict:
    index_map = {}

    for idx, value in enumerate(transaction):
        if value in index_map:
            index_map[value].append(idx)
        else:
            index_map[value] = [idx]

    # keep only duplicates
    duplicates = {k: v for k, v in index_map.items() if len(v) > 1}

    return duplicates


# Example usage
transaction = [101,303,405,101,904,402,784,904,103,433]
print(find_duplicate_indices(transaction))
```

```python
# 👉 Fetch max value from list in python without using inbuilt function

num = [10,25,50,60,87,100,78]

#using inbuilt function
#max(num)

def get_maxValue(num):
    
    max_val = num[0]
    
    for n in num:
        if n > max_val:
            max_val = n
    
    return max_val
    

get_maxValue(num)
```

```python
# 👉 Find SECOND LARGEST element (No inbuilt)
num = [10,25,30,43,53,22,21,90]

def second_largest(nums):
    if len(nums) < 2:
        return None

    largest = second = float('-inf')

    for n in nums:
        if n > largest:
            second = largest
            largest = n
        elif largest > n > second:
            second = n

    return second


print(second_largest(num))
```
