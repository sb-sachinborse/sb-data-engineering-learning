```python
from pyspark.sql import SparkSession
import os
os.environ["SPARK_LOCAL_IP"] = "127.0.0.1"
from pyspark.sql.functions import (
    col, when, sum, avg, count,
    max, min, desc, asc
)
```

```python
spark = SparkSession.builder\
        .appName("sparksql")\
        .getOrCreate()

print(spark)
print(f"spark_version:",spark.version)
        
```

```python
#read csv
df = spark.read.format('csv')\
    .option('inferSchema','true')\
    .option('header','true')\
    .option('path','datasets/operations_management.csv')\
    .load()
```

```python
df.printSchema()
```

```python
df.show(5)
```

```python
data = df.select("industry","value")\
    .filter(col("value") > 10000)\
    .orderBy(desc("value"))

data.printSchema()
```

```python
data.show(5)
```

#### spark sql

```python
data.createOrReplaceTempView("df")
```

```python
spark.sql("""
    SELECT 
        industry,
        value
    FROM df
    WHERE value > 10000
""").show(5)
```
