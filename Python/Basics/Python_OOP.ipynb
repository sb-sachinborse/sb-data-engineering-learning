## 🐍 Python OOP for Data Engineers (Complete Guide)

#### 1️⃣ Why OOP for Data Engineers?
Without OOP (Bad)
- Hard-coded logic
- Repeated code
-Difficult to maintain pipelines

With OOP (Good)
- Reusable ETL component
- Clear separation of concerns
- Easy scaling & testing

#### 2️⃣ Class & Object (Foundation)

- Class → Blueprint
- Object → Instance of class


```python
class DataEngineer:
    def work(self):
        print("Building data pipelines")

de = DataEngineer()
de.work()
```

#### 3️⃣ Constructor (__init__)
🔹 Used to initialize object data

```python
class Employee:
    def __init__(self, name, role):
        self.name = name
        self.role = role

    def show(self):
        print(self.name, self.role)

e1 = Employee("Krishna", "Data Engineer")
e1.show()
```
