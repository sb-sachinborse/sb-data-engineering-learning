#### Dataframe with header and proper schema

```python
df2 = spark.read.csv('datasets/emp_1.csv', header=True,inferSchema=True)
df2.show()
```

```python
df2.printSchema()
```

```python
df3 = spark.read.format("csv").options(header=True,inferSchema=True).load('datasets/emp_1.csv')
df3.show()
```
