# Group By 

It allows you to perform aggreate functions on groups of row , rahter than on individual rows, enabling you  to summerize data and generate aggreate statisic

#### aggregate function
- count() : Return the number of rows of each group
- mean()  : Return the mean of values for each group
- max()   : Return the maximum of values for each group
- min()   : Return the minumun of vlaues for each group
- sum()   : Return the total for values for each group
- avg()   : Return the average for values for each group
- agg()   : using groupBy() agg() function we can calculate more than one aggreate at time. 

```python
df = spark.read.csv("datasets/emp_1.csv", header=True, inferSchema=True)
df.show()
```

```python
#based on each address total number of salary
df.groupBy('address').sum('salary').show()
```

```python
#total number of count
df.groupBy('address').count().show()
```

```python
#
df.groupBy('address','emp_id').sum('salary').show()
```

```python
#average salary by each address
df.groupBy('address').avg('salary').show()
```

```python
#max salary of each address
df.groupBy('address').max('salary').show()
```

```python
#min salary of each address
df.groupBy('address').min('salary').show()
```
