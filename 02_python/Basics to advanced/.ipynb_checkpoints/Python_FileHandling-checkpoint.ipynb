## 🐍 File Handling for Data Engineers

##### CSV • JSON • Parquet (Deep Dive)

#### 1️⃣ File Handling Basics
- Files are external data sources
- Python supports file I/O using open()
- Modes:
    - r → read
    - w → write (overwrite)
    - a → append

```python
file = open("data.txt", "r")
print(file.read())
file.close()

# Always close file or use with
```

#### 2️⃣ CSV FILE HANDLING
🔹 What is CSV?
- Comma-separated values
- Row-based structured data
- Common in data ingestion

##### .1 Reading CSV (Basic – open())

```python
with open("employees.csv", "r") as file:
    for line in file:
        print(line.strip())
```

##### 2 CSV Module (Recommended)

```python
import csv

with open("employees.csv", newline="") as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

##### 3 Reading CSV as Dictionary

```python
import csv

with open("employees.csv", newline="") as file:
    reader = csv.DictReader(file)
    for row in reader:
        print(row["name"], row["salary"])
```

##### 4 Writing CSV

```python
import csv

data = [
    ["id", "name", "salary"],
    [1, "Sachin", 50000],
    [2, "Amit", 60000]
]

with open("output.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerows(data)
```

##### 5 Writing CSV Using DictWriter

```python
import csv

rows = [
    {"id": 1, "name": "Sachin"},
    {"id": 2, "name": "Amit"}
]

with open("out.csv", "w", newline="") as file:
    writer = csv.DictWriter(file, fieldnames=["id", "name"])
    writer.writeheader()
    writer.writerows(rows)
```

##### 🧠 Data Engineer CSV Cleaning Example

```python
import csv

clean = []

with open("raw.csv") as file:
    reader = csv.DictReader(file)
    for row in reader:
        if row["salary"].isdigit():
            row["salary"] = int(row["salary"])
            clean.append(row)

print(clean)
```

#### 3️⃣ JSON FILE HANDLING

###### 🔹 What is JSON?

- Key-value structured data
- Nested
- Used in APIs, configs

##### 1 Read JSON File

```python
import json

with open("data.json") as file:
    data = json.load(file)

print(data)
```

##### 2 Write JSON File

```python
import json

data = {"name": "Sachin", "role": "Data Engineer"}

with open("out.json", "w") as file:
    json.dump(data, file, indent=4)
```

##### 3 Read JSON Lines (Big Data)

```python
import json

with open("events.json") as file:
    for line in file:
        record = json.loads(line)
        print(record)
```

##### 4 Nested JSON Access

```python
data = {
    "user": {
        "id": 1,
        "skills": ["Python", "SQL"]
    }
}

print(data["user"]["skills"][0])
```

##### 🧠 API → JSON → Clean Example

```python
import json

with open("api.json") as f:
    data = json.load(f)

clean = []
for r in data:
    clean.append({
        "id": r["id"],
        "city": r["address"]["city"]
    })

print(clean)
```

#### 4️⃣ CSV vs JSON (INTERVIEW TABLE)

| Feature    | CSV       | JSON          |
| ---------- | --------- | ------------- |
| Structure  | Tabular   | Nested        |
| Size       | Smaller   | Larger        |
| Read Speed | Fast      | Medium        |
| Use Case   | Flat data | APIs, configs |
| Schema     | Fixed     | Flexible      |


#### REAL ETL PIPELINE (CSV → JSON)

```python
import csv, json

with open("input.csv") as csvfile:
    reader = csv.DictReader(csvfile)
    data = list(reader)

with open("output.json", "w") as jsonfile:
    json.dump(data, jsonfile, indent=2)
```

#### 6️⃣ PARQUET AWARENESS (VERY IMPORTANT)
What is Parquet?

- Columnar storage format
- Compressed
- Schema-based
- Designed for big data

🔹 Why Data Engineers Use Parquet?
- Faster reads
- Smaller size
- Optimized for Spark
