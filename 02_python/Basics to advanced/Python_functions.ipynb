## 🐍 Python Functions

```python
# 1️⃣ What is a Function?
# ✅ Definition

# A function is a block of reusable code that performs a specific task.

def function_name():
    # code
```

```python
#2️⃣ Simple Function (No Parameters, No Return

def greet():
    print("Hello, Welcome to Python!")

greet()
```

```python
#3️⃣ Function with Parameters

def greet(name):
    print(f"Hello {name}")

greet("Ram")
```

```python
#4️⃣ Function with Multiple Parameters

def add(a, b):
    print(a + b)

add(10, 20)
```

```python
#5️⃣ Function with Return Value

def add(a, b):
    return a + b

result = add(5, 3)
print(result)

# 🧠 Key Point
# return sends value back to caller
# Stops function execution
```

```python
#6️⃣ Function with Default Parameters

def greet(name="Guest"):
    print(f"Hello {name}")

greet()
greet("Sachin")
```

```python
#8️⃣ Variable Length Arguments (*args)

def add(*numbers):
    total = 0
    for n in numbers:
        total += n
    return total

print(add(1, 2, 3, 4))
```

```python
#9️⃣ Keyword Variable Arguments (**kwargs)
#**kwargs syntax allows a function to accept a variable number of keyword (named) arguments. 
# These arguments are collected into a dictionary within the function, 
# where the keys are the argument names and the values are the argument values. 

def student(**details):
    for key, value in details.items():
        print(key, ":", value)

student(name="Sachin", age=25, city="Pune")

```

```python
#🔟 Function Inside Function (Nested Function)

def outer():
    def inner():
        print("Inner Function")
    inner()

outer()

```

```python
#1️⃣1️⃣ Lambda Function (Anonymous Function)

square = lambda x: x * x
print(square(5))

#📌 With multiple arguments

add = lambda a, b: a + b
print(add(10, 20))
```

```python
# 1️⃣2️⃣ Function with List Input

def total(nums):
    return sum(nums)

print(total([1, 2, 3, 4]))
```

```python
# 1️⃣3️⃣ Recursion (Function Calling Itself)

def factorial(n):
    if n == 1:
        return 1
    return n * factorial(n - 1)

print(factorial(5))
```
