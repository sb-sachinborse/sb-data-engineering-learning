## 🐍 Python – Input from User (Complete Guide)

```python
# 1️⃣ Basic input()
# ✅ What it does

# Takes input from keyboard
# Always returns a string

name = input("Enter your name: ")
print(name)
```

```python
#2️⃣ Input + Type Checking
age = input("Enter your age: ")
print(type(age))   # <class 'str'>
```

```python
# 3️⃣ Type Casting User Input
# Convert to Integer
age = int(input("Enter your age: "))
print(age + 5)

#Convert to Float
salary = float(input("Enter salary: "))
print(salary)
```

```python
# 4️⃣ Multiple Inputs in One Line
# 📌 Using split()

a, b = input("Enter two numbers: ").split()
print(a, b)

#With Type Conversion
a, b = map(int, input("Enter two numbers: ").split())
print(a + b)

```

```python
# 5️⃣ Taking List Input from User
# 📌 Space-separated list
nums = list(map(int, input("Enter numbers: ").split()))
print(nums)
```

```python
#6️⃣ Taking Tuple Input
values = tuple(input("Enter values: ").split())
print(values)
```

```python
# 7️⃣ Taking Dictionary Input (Simple Way)
student = {}

student["name"] = input("Enter name: ")
student["age"] = int(input("Enter age: "))

print(student)
```

```python
# 8️⃣ Boolean Input
# ⚠ No direct boolean input — handle manually

status = input("Enter True or False: ").lower()

is_active = status == "true"
print(is_active)
```

```python
# 9️⃣ Input in Loop
# 📌 Using for

marks = []

for i in range(3):
    marks.append(int(input("Enter mark: ")))

print(marks)
```

```python
#📌 Using while
nums = []

while True:
    n = int(input("Enter number (-1 to stop): "))
    if n == -1:
        break
    nums.append(n)

print(nums)
```

```python
# 🔟 Input Validation (Very Important)
# 📌 Handle Wrong Input

try:
    age = int(input("Enter age: "))
    print("Age:", age)
except ValueError:
    print("Invalid input! Enter number only.")
```

```python
# 1️⃣1️⃣ Password Input (Hidden)
import getpass

password = getpass.getpass("Enter password: ")
print("Password entered")
```

```python
#1️⃣2️⃣ Command Line Input (Advanced)
import sys

name = sys.argv[1]
print(name)

# Run:
# python app.py Ram
```
