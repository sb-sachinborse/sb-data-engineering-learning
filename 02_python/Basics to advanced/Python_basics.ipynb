# 🐍 Python Basics – Coding Examples

#### 1️⃣ Variables in Python
✅ What is a Variable?

A variable stores data in memory.
Python is dynamically typed, so you don’t declare the type explicitly.

```python
name = "John"
age = 25
salary = 45000.50
is_active = True

print(name)
print(age)
print(salary)
print(is_active)
```

#### 🧠 Key Points

- No int, float, string keywords
- Type is decided at runtime
- Variables can be reassigned

#### 2️⃣ Type Casting (Type Conversion)
✅ What is Type Casting?
Converting one data type into another.

```python
a = "100"
b = 20

# Convert string to int
a = int(a)

print(a + b)
```

```python
x = 10
print(float(x))   # int → float

y = 12.5
print(int(y))     # float → int

z = 50
print(str(z))     # int → string
```

#### 4️⃣ If – Else Condition
✅ Used for decision making

```python
age = 18

if age >= 18:
    print("Eligible to vote")
else:
    print("Not eligible")
```

🔁 if – elif – else

```python
marks = 75

if marks >= 90:
    print("Grade A")
elif marks >= 60:
    print("Grade B")
else:
    print("Grade C")
```

#### 5️⃣ For Loop (Normal Loop)
✅ Used when number of iterations is known

```python
for i in range(1, 6):
    print(i)
```

```python
# 📌 Print Even Numbers
for i in range(2, 11, 2):
    print(i)
```

#### 6️⃣ For Each Loop (Python Style)
👉 In Python, for-each is just for

```python
#Example (List)
fruits = ["apple", "banana", "mango"]

for fruit in fruits:
    print(fruit)
```

```python
# 📌 Dictionary Loop
student = {"name": "Sachin", "age": 25}

for key, value in student.items():
    print(key, ":", value)
```

#### 7️⃣ While Loop
✅ Used when condition-based looping is required

```python
count = 1

while count <= 5:
    print(count)
    count += 1

# 📌 Infinite Loop (⚠️ Careful)
# while True:
#     print("Hello")
```

#### 8️⃣Switch Case (Python Alternative)
Python does not have switch, but from Python 3.10+ we have:

```python
day = 3

match day:
    case 1:
        print("Monday")
    case 2:
        print("Tuesday")
    case 3:
        print("Wednesday")
    case _:
        print("Invalid day")
```
