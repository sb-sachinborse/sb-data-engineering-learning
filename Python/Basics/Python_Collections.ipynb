# Python Collections Deep Dive
### List / Tuple / Set / Dictionar

## 1. List
- Ordered
- Mutable
- Allows duplicates
- Indexed
- Most used in ETL pipelines

```python
numbers = [10, 20, 30, 40]
```

```python
##🔧 Common List Operations

nums = [1, 2, 3]

nums.append(4)
nums.extend([5, 6])
nums.insert(1, 100)

nums.remove(3)
nums.pop()
nums.clear()

```

```python
#🔁 Looping

for n in nums:
    print(n)
```

```python
# 🧠 Data Engineer Example
# Cleaning numeric data

raw_data = ["10", "20", None, "30", ""]

clean_data = []
for x in raw_data:
    if x and x.isdigit():
        clean_data.append(int(x))

print(clean_data)
```

## 2. Tuple
- Ordered
- Immutable
- Allows duplicates
- Faster than list
- Used for fixed schema data

```python
# Tuple examples

record = (101, "Sachin", 50000)

#🔍 Accessing
print(record[0])

#❌ Not Allowed
# record[1] = "Amit"  ❌
```

```python
#🧠 Data Engineer Example
#Fixed schema row
employee = (101, "Sachin", "Data Engineer", 75000)
```

## 3. Set
- Unordered
- Mutable
- No duplicates
- Very fast lookup
- Used for deduplication

```python
ids = {1, 2, 3, 3, 4}
print(ids)
```

```python
# 🔧 Set Operations
a = {1, 2, 3}
b = {3, 4, 5}

print(a | b)   # union
print(a & b)   # intersection
print(a - b)   # difference
```

```python
# 🧠 Data Engineer Example
# Deduplicate records

user_ids = [101, 102, 103, 101, 104]

unique_ids = set(user_ids)
print(unique_ids)
```

### 4. DICTIONARY
- Key–Value pairs
- Mutable
- Keys must be unique
- Backbone of JSON, APIs, NoSQL

```python
employee = {
    "id": 101,
    "name": "Ram",
    "role": "Data Engineer"
}
```

```python
#🔧 Dictionary Operations
employee["salary"] = 75000
employee["role"] = "Senior DE"

employee.pop("salary")
```

```python
# 🔁 Loop Dictionary

for key, value in employee.items():
    print(key, value)
```

```python
# 🧠 Data Engineer Example
# JSON record processing

record = {
    "id": 101,
    "name": "Krishna",
    "skills": ["Python", "SQL", "Spark"]
}

if "skills" in record:
    print(record["skills"])
```

🔥 COMPARISON TABLE (IMPORTANT)

| Feature    | List     | Tuple        | Set           | Dict         |
| ---------- | -------- | ------------ | ------------- | ------------ |
| Ordered    | ✅        | ✅            | ❌             | ✅            |
| Mutable    | ✅        | ❌            | ✅             | ✅            |
| Duplicates | ✅        | ✅            | ❌             | ❌ (keys)     |
| Indexing   | ✅        | ✅            | ❌             | Keys         |
| Speed      | Medium   | Fast         | Very Fast     | Fast         |
| Use-case   | ETL data | Fixed schema | Deduplication | JSON, config |


💼 Data Engineer Real-World Mapping

| Task              | Collection |
| ----------------- | ---------- |
| Raw CSV rows      | List       |
| DB row schema     | Tuple      |
| Remove duplicates | Set        |
| JSON / API data   | Dictionary |
| Spark row → dict  | Dict       |
| Column values     | List       |


### 🚀 Combined Example (ETL Style)

```python
raw_records = [
    {"id": 1, "city": "Pune"},
    {"id": 2, "city": "Mumbai"},
    {"id": 1, "city": "Pune"}
]

unique = {}
for r in raw_records:
    unique[r["id"]] = r

clean_records = list(unique.values())
print(clean_records)
```
