### SELECT function in pyspark

select() function is used to select single , multiple, column by index, all columns from the list and the nested columns from a DataFrame

```python
from pyspark.sql import *

import os
os.environ["SPARK_LOCAL_IP"] = "127.0.0.1"

spark = SparkSession.builder \
    .appName("spark_lab_3") \
    .getOrCreate()

spark
```

```python
df = spark.read.csv("datasets/emp_1.csv",header=True,inferSchema=True)
df.show()
```

```python
df.select("emp_id","salary").show()
```

```python
#another way
df.select(df.emp_id,df.salary).show()
```

```python
df.select(df.columns[3:6]).show()  
```

## withColumn() and  withColumnRenamed()

pyspark withColumn() is a transformation function of DataFrame which is used to change the value , convert the 
datatype of an existing column , create a new column and many more

withColumnRenamed() -> use pyspark withColumnRenamed() to rename a DataFrame column, we often need to rename 
one column or multiple(or all) columns on pyspark DataFrame
1. create DataFrame
2. add new column based on existing column
3. add new column
4. change datatype
5. update teh valud of existing coloumns

```python
data = spark.read.csv('datasets/emp_1.csv', header=True)
data.show()
```

```python
#add new column based on existing column

data.withColumn('salary1',data.salary*10).show() 
```

```python
#add new column based on constant value
from pyspark.sql.functions import lit
data.withColumn('country',lit('usa')).show()
```

```python
data.withColumn('increment',lit(0)).show()
```

```python
#update existing schema
data.printSchema()
```

```python
from pyspark.sql.functions import col
updatedData = data.withColumn('salary', col('salary').cast('Integer'))
updatedData.printSchema()
```

```python
#column rename
data.withColumnRenamed('loc','location').show()
```

### Filter Function

#### filter function similar to where function in SQL


```python
#filter all address from india
df.filter(df.address == 'india').show()
```

```python
#addresss != india
df.filter(df.address != 'india').show()
```

```python
#where name = manish and salary should equal to 1000
df.filter((df.name == 'manish') & (df.salary == 10000)).show()
```

```python
#show the record where slaray is less than 30000
df.filter(df.salary < 30000).show()
```

```python
#name starting with r
df.filter(df.name.startswith("r")).show()
```

```python
#name ending with 'l'
df.filter(df.name.endswith('l')).show()
```

### based on expression

```python
# filter data where name have 'ul'
df.filter(df.name.like('%ul%')).show()
```

### Distinct() and dropDuplicates()

- distinct() transform is used to drop/remove the duplicates row(all columns) from dataFrame
- dropDuplicates() is used to drop row based on selected(one or multiple) columns

##### interview question
  -Diffrence between distinct and dropDuplicates function

```python
df1 = df.distinct()
df1.show()
```

```python
df2= df.dropDuplicates(['emp_id','salary'])
df2.show()
```

## sort and orderBy function

- we can use either sort() or orderBy() function of pyspark DataFrame to sort DataFrame by ascending or descending order based on single or multiple columns

```python
#df.sort(df.salary).show()

df.sort('salary').show()
```

```python
df.orderBy(df.salary).show()
```

```python
df.sort(df.name,df.salary).show()
```

```python
#deceding order
df.sort(df.salary.desc()).show()
```

```python
df.orderBy(df.salary.desc()).show()
```
