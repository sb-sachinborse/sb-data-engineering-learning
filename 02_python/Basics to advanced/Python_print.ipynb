### 🐍 Python print() – All Methods & Techniques

```python
#1️⃣ Basic print()
print("Hello World")
print(10)
print(10.5)
print(True)
```

```python
#2️⃣ Printing Multiple Values
#📌 Using commas

name = "John"
age = 25

print(name, age)
```

```python
# 3️⃣ sep (Separator)
# 📌 Change separator
print("Python", "Java", "Go", sep=", ")
```

```python
# 4️⃣ end (Line Ending)
# 📌 Prevent new line

print("Hello", end=" ")
print("World")
```

```python
#5️⃣ Print New Line Explicitly
print("Hello\nWorld")
```

```python
# 6️⃣ String Concatenation (+)
#⚠ Works only with strings

name = "Krishna"
print("My name is " + name)
```

```python
#7️⃣ Type Casting in Print

age = 25
print("Age is " + str(age))
```

```python
# 8️⃣ f-Strings (🔥 Best & Modern Way)
# 📌 Recommended

name = "Sachin"
age = 25

print(f"My name is {name} and age is {age}")

#📌 Expressions inside print
a = 10
b = 20
print(f"Sum is {a + b}")

```

```python
#9️⃣ format() Method

name = "Ram"
age = 35

print("My name is {} and age is {}".format(name, age))


#📌 Index-based
print("Age: {1}, Name: {0}".format("Ram", 25))

```

```python
# 🔟 % Formatting (Old Style)

name = "Krishna"
age = 25

print("My name is %s and age is %d" % (name, age))
```

```python
#1️⃣2️⃣ Printing Data Structures

#📌 List
nums = [1, 2, 3]
print(nums)

#📌 Dictionary
student = {"name": "Ram", "age": 25}
print(student)

```

```python
# 1️⃣5️⃣ Printing Without Quotes
print('Python')
print("Python")
print("""Python""")
```

```python
#Printing Special Characters
print("Tab\tSpace")
print("Backslash \\")
print("Quote \"Python\"")
```

```python
#Printing to File

file = open("output.txt", "w")
print("Hello File", file=file)
file.close()
```
