#### 1️⃣ String Basics
- String is immutable
- Sequence of characters
- Indexed
- Can use single, double, triple quotes

```python
s = "Data Engineer"
```

#### 2️⃣ String Indexing & Slicing

```python


s = "Python"

print(s[0])     # P
print(s[-1])    # n
print(s[1:4])   # yth
print(s[:3])    # Pyt
print(s[::2])   # Pto
```

#### 🧠 Data Engineer Use
Extract columns, prefixes, suffixes

#### 3️⃣ Case Conversion Methods

```python


text = "PyThOn"

print(text.lower())
print(text.upper())
print(text.title())
print(text.capitalize())
print(text.swapcase())
```

#### 4️⃣ Strip / Trim Methods (Very Important)

data = "   Pune   "

print(data.strip())
print(data.lstrip())
print(data.rstrip())

#🧠 Used for cleaning CSV values

#### 5️⃣ Replace & Remove Characters

```python
text = "2024-01-15"

print(text.replace("-", "/"))
```

```python
#Remove unwanted characters:

s = "₹45,000"
print(s.replace("₹", "").replace(",", ""))
```

#### 6️⃣ Find, Index, Count

```python
s = "data engineering data"

print(s.find("data"))
print(s.rfind("data"))
print(s.count("data"))
```

#### 7️⃣ Startswith / Endswith

```python
file = "data.csv"

print(file.startswith("data"))
print(file.endswith(".csv"))

#🧠 File validation in ETL
```

#### 8️⃣ Split & Join (🔥 MOST IMPORTANT)

```python
#Split

line = "101,Ram,Mumbai,75000"

parts = line.split(",")
print(parts)
```

```python
#Join

values = ["2026", "01", "03"]
date = "-".join(values)
print(date)

#🧠 CSV & log parsing
```

#### 9️⃣ Check String Content

```python
#🧠 Data validation

s = "12345"

print(s.isdigit())
print(s.isalpha())
print(s.isalnum())
print(s.isspace())
print(s.islower())
print(s.isupper())
```

#### 1️⃣4️⃣ Reverse String

```python
s = "Python"
print(s[::-1])
```

#### 1️⃣5️⃣ Sorting Characters

```python
s = "dcba"
print("".join(sorted(s)))
```

#### 1️⃣6️⃣ Removing Duplicates

```python
s = "aabbcc"
print("".join(set(s)))
```

#### 1️⃣7️⃣ Preserve Order While Removing Duplicates

```python
s = "aabbcc"

result = ""
for c in s:
    if c not in result:
        result += c

print(result)
```

#### 1️⃣8️⃣ Count Character Frequency

```python
s = "data"

freq = {}
for c in s:
    freq[c] = freq.get(c, 0) + 1

print(freq)

#🧠 Used in data profiling
```

#### 1️⃣9️⃣ Regex String Manipulation (Data Engineer Level)

```python
import re

text = "Order ID: 123, Amount: 4500"

numbers = re.findall(r"\d+", text)
print(numbers)
```

#### 2️⃣0️⃣ Cleaning Real CSV Line (ETL Example)

```python
line = " 101 , Ram , Pune , ₹45,000 "

clean = line.strip().replace("₹", "").replace(",", "")
cols = [c.strip() for c in clean.split()]

print(cols)
```
